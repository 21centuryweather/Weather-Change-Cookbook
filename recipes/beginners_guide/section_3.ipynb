## Handle a datetime column

Climate CSV files usually have a 'time' column stored as a string. To do time-based operations, you need to tell pandas it's a date, otherwise it doesn’t know (silly pandas!).

```python
# Option 2: convert after loading
df["time"] = pd.to_datetime(df["time"])

# Set time as the index for easy slicing
df = df.set_index("time")
print(df.head())
```

:::{note} `parse_dates = pd.read_csv(file, parse_dates=["time"])`

Tells pandas to interpret the named column(s) as datetime objects rather than plain strings. This unlocks time-based filtering and resampling.
:::
